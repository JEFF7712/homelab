# Category drilldown

```python
import os, sqlite3
from pathlib import Path
import pandas as pd
import seaborn as sns

DB = Path(os.environ.get('PREDMARKBOT_RESEARCH_DB',
    str(Path.home() / '.local/share/predmarkbot/research.db')))
conn = sqlite3.connect(f'file:{DB}?mode=ro', uri=True)
```

```python
df = pd.read_sql('SELECT * FROM bucket_stats WHERE category != "ALL"', conn)
g = sns.FacetGrid(df[df.horizon=='T-6h'], col='category', col_wrap=3, height=3, sharey=True)
g.map_dataframe(sns.scatterplot, x='bucket_lo', y='bias_bps')
g.set_axis_labels('bucket lo (¢)', 'bias (bps)')
```
