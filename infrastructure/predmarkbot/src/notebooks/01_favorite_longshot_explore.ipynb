# Favorite-longshot exploration

Read-only view of `research.db`. Mutate via the CLI (`predmarkbot research analyze`).

```python
import os, sqlite3
from pathlib import Path
import pandas as pd

DB = Path(os.environ.get('PREDMARKBOT_RESEARCH_DB',
    str(Path.home() / '.local/share/predmarkbot/research.db')))
conn = sqlite3.connect(f'file:{DB}?mode=ro', uri=True)
print(DB, 'connected')
```

```python
bucket_stats = pd.read_sql('SELECT * FROM bucket_stats', conn)
bucket_stats.head()
```

```python
# Bias curve for ALL at T-6h
subset = bucket_stats[(bucket_stats.horizon=='T-6h') & (bucket_stats.category=='ALL')].sort_values('bucket_lo')
subset[['bucket_lo','n_markets','realized_rate','expected_rate','bias_bps','p_value']]
```
